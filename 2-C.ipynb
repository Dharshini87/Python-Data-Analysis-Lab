```python
import pandas as pd
from io import StringIO, BytesIO

print("=" * 60)
print("EXPERIMENT 3: READING DATA FROM TEXT FILES, EXCEL AND WEB")
print("=" * 60)

# ---------------------------------------------------------
# 1. READ DATA FROM TEXT/CSV FILE
# ---------------------------------------------------------

csv_data = """App,Category,Rating,Reviews
Photo Editor,ART_AND_DESIGN,4.1,159
Coloring Book,ART_AND_DESIGN,3.9,967
U Launcher Lite,ART_AND_DESIGN,4.7,87510
Sketch Draw Paint,ART_AND_DESIGN,4.5,215644
Pixel Draw,ART_AND_DESIGN,4.3,967
"""

text_df = pd.read_csv(StringIO(csv_data))

print("\n1. DATA FROM TEXT/CSV FILE")
print("-" * 40)
print(text_df.head())

# ---------------------------------------------------------
# 2. READ DATA FROM EXCEL FILE
# ---------------------------------------------------------

excel_data = {
    "Product": ["Laptop", "Smartphone", "Tablet", "Headphones"],
    "Price": [1000, 800, 500, 100],
    "Quantity": [5, 8, 10, 15]
}

excel_df_original = pd.DataFrame(excel_data)

# Create Excel file in memory
excel_buffer = BytesIO()

with pd.ExcelWriter(excel_buffer, engine="openpyxl") as writer:
    excel_df_original.to_excel(
        writer,
        sheet_name="Sheet1",
        index=False
    )

excel_buffer.seek(0)

# Read Excel file
excel_df = pd.read_excel(
    excel_buffer,
    sheet_name="Sheet1"
)

print("\n2. DATA FROM EXCEL FILE")
print("-" * 40)
print(excel_df.head())

# ---------------------------------------------------------
# 3. READ DATA FROM WEB
# ---------------------------------------------------------

url = "https://raw.githubusercontent.com/cs109/2014_data/master/countries.csv"

web_df = pd.read_csv(url)

print("\n3. DATA FROM WEB")
print("-" * 40)
print(web_df.head())

# ---------------------------------------------------------
# 4. DISPLAY DATASET INFORMATION
# ---------------------------------------------------------

print("\n4. TEXT/CSV DATA INFORMATION")
print("-" * 40)
text_df.info()

print("\n5. EXCEL DATA INFORMATION")
print("-" * 40)
excel_df.info()

print("\n6. WEB DATA INFORMATION")
print("-" * 40)
web_df.info()

# ---------------------------------------------------------
# 5. CHECK MISSING VALUES
# ---------------------------------------------------------

print("\n7. MISSING VALUES IN TEXT/CSV DATA")
print("-" * 40)
print(text_df.isnull().sum())

print("\n8. MISSING VALUES IN EXCEL DATA")
print("-" * 40)
print(excel_df.isnull().sum())

print("\n9. MISSING VALUES IN WEB DATA")
print("-" * 40)
print(web_df.isnull().sum())

# ---------------------------------------------------------
# 6. HANDLE MISSING VALUES
# ---------------------------------------------------------

text_df = text_df.ffill()
excel_df = excel_df.bfill()
web_df = web_df.dropna()

print("\n10. MISSING VALUES HANDLED SUCCESSFULLY")

# ---------------------------------------------------------
# 7. DISPLAY PROCESSED DATA
# ---------------------------------------------------------

print("\n11. PROCESSED TEXT/CSV DATA")
print("-" * 40)
print(text_df.head())

print("\n12. PROCESSED EXCEL DATA")
print("-" * 40)
print(excel_df.head())

print("\n13. PROCESSED WEB DATA")
print("-" * 40)
print(web_df.head())

# ---------------------------------------------------------
# 8. SAVE PROCESSED DATA
# ---------------------------------------------------------

processed_csv = text_df.to_csv(index=False)

processed_excel_buffer = BytesIO()

with pd.ExcelWriter(
    processed_excel_buffer,
    engine="openpyxl"
) as writer:
    excel_df.to_excel(
        writer,
        sheet_name="ProcessedData",
        index=False
    )

print("\n14. PROCESSED DATA")
print("-" * 40)
print("CSV data processed successfully.")
print("Excel data processed successfully.")

# ---------------------------------------------------------
# 9. FINAL RESULT
# ---------------------------------------------------------

print("\n" + "=" * 60)
print("RESULT")
print("=" * 60)

print(
    "The experiment successfully demonstrated reading and "
    "processing data from text/CSV files, Excel spreadsheets, "
    "and web-based sources using Pandas."
)

print(
    "Missing values were handled successfully and the "
    "processed data was prepared for export."
)

print("=" * 60)
print("EXPERIMENT 3 COMPLETED SUCCESSFULLY")
print("=" * 60)
```
