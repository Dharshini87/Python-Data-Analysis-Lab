```python
import pandas as pd
from io import StringIO

# --------------------------------------------------
# 1. Create and load dataset
# --------------------------------------------------

csv_data = """Name,Age,Marks,Attendance,Department
Arun,20,85,90,CSE
Priya,21,92,95,ECE
Rahul,19,76,80,CSE
Divya,22,88,88,IT
Karthik,20,95,92,ECE
Meena,21,79,85,CSE
Arun,20,85,90,CSE
Vijay,,90,94,IT
"""

df = pd.read_csv(StringIO(csv_data))

print("Original DataFrame:")
print(df)

# --------------------------------------------------
# 2. Display first and last rows
# --------------------------------------------------

print("\nFirst 5 Rows:")
print(df.head())

print("\nLast 5 Rows:")
print(df.tail())

# --------------------------------------------------
# 3. Data types and information
# --------------------------------------------------

print("\nData Types:")
print(df.dtypes)

print("\nDataFrame Information:")
df.info()

# --------------------------------------------------
# 4. Summary statistics
# --------------------------------------------------

print("\nSummary Statistics:")
print(df.describe())

# --------------------------------------------------
# 5. Identify missing values
# --------------------------------------------------

print("\nMissing Values:")
print(df.isnull().sum())

# --------------------------------------------------
# 6. Handle missing values
# --------------------------------------------------

df['Age'] = df['Age'].fillna(df['Age'].median())

print("\nAfter Handling Missing Values:")
print(df)

# --------------------------------------------------
# 7. Create a new column
# --------------------------------------------------

df['Marks_Double'] = df['Marks'] * 2

print("\nNew Column - Marks_Double:")
print(df[['Name', 'Marks', 'Marks_Double']])

# --------------------------------------------------
# 8. Create a Series and perform operations
# --------------------------------------------------

series = df['Marks']

print("\nMarks Series:")
print(series)

print("\nSeries + 10:")
print(series + 10)

# --------------------------------------------------
# 9. Filter using multiple conditions
# --------------------------------------------------

filtered_df = df[
    (df['Marks'] > 80) &
    (df['Attendance'] > 85)
]

print("\nFiltered DataFrame:")
print(filtered_df)

# --------------------------------------------------
# 10. Grouping and aggregation
# --------------------------------------------------

grouped = df.groupby('Department')['Marks'].mean()

print("\nAverage Marks by Department:")
print(grouped)

# --------------------------------------------------
# 11. Sorting
# --------------------------------------------------

df_sorted = df.sort_values(
    by='Marks',
    ascending=False
)

print("\nSorted DataFrame:")
print(df_sorted)

# --------------------------------------------------
# 12. Boolean masking
# --------------------------------------------------

median_marks = df['Marks'].median()

masked_df = df[df['Marks'] > median_marks]

print("\nStudents Above Median Marks:")
print(masked_df)

# --------------------------------------------------
# 13. Remove duplicates and missing values
# --------------------------------------------------

df_cleaned = df.drop_duplicates()
df_cleaned = df_cleaned.dropna()

print("\nAfter Removing Duplicates and Missing Values:")
print(df_cleaned)

# --------------------------------------------------
# 14. Create subset DataFrame
# --------------------------------------------------

subset_df = df_cleaned[
    ['Name', 'Department', 'Marks']
]

print("\nSelected Columns:")
print(subset_df)

# --------------------------------------------------
# 15. Save result in memory / display result
# --------------------------------------------------

print("\nFiltered DataFrame to be Saved:")
print(subset_df)

# --------------------------------------------------
# 16. Summary statistics
# --------------------------------------------------

print("\nTotal Marks:", df_cleaned['Marks'].sum())
print("Mean Marks:", df_cleaned['Marks'].mean())
print("Standard Deviation:", df_cleaned['Marks'].std())

print("\nExperiment 2 completed successfully!")
```
