```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

print("=" * 70)
print("3-C: MULTIPLE REGRESSION ANALYSIS")
print("UCI AND PIMA INDIANS DIABETES DATASETS")
print("=" * 70)

# ------------------------------------------------------------
# 1. Load Diabetes Dataset
# ------------------------------------------------------------

url = "https://raw.githubusercontent.com/jbrownlee/Datasets/master/pima-indians-diabetes.data.csv"

columns = [
    "Pregnancies",
    "Glucose",
    "BloodPressure",
    "SkinThickness",
    "Insulin",
    "BMI",
    "DiabetesPedigreeFunction",
    "Age",
    "Outcome"
]

pima_diabetes = pd.read_csv(
    url,
    names=columns
)

# Working copy for UCI comparison
uci_diabetes = pima_diabetes.copy()

print("\n1. DATASET INFORMATION")
print("-" * 50)

print("Pima Dataset Shape:", pima_diabetes.shape)
print("UCI Working Dataset Shape:", uci_diabetes.shape)

print("\nFirst 5 rows:")
print(pima_diabetes.head())

# ------------------------------------------------------------
# 2. Select Features and Target
# ------------------------------------------------------------

features = [
    "Glucose",
    "BloodPressure",
    "Age"
]

target = "BMI"

print("\n2. VARIABLES SELECTED")
print("-" * 50)

print("Independent Variables:", features)
print("Dependent Variable:", target)

# ------------------------------------------------------------
# 3. Handle Invalid Values
# ------------------------------------------------------------

for data in [uci_diabetes, pima_diabetes]:

    for column in [
        "Glucose",
        "BloodPressure",
        "BMI"
    ]:

        data[column] = data[column].replace(
            0,
            np.nan
        )

        data[column] = data[column].fillna(
            data[column].median()
        )

# ------------------------------------------------------------
# 4. Multiple Regression Function
# ------------------------------------------------------------

def multiple_regression_analysis(
    data,
    dataset_name
):

    X = data[features]
    y = data[target]

    # Split data into training and testing sets
    X_train, X_test, y_train, y_test = train_test_split(
        X,
        y,
        test_size=0.20,
        random_state=42
    )

    # Create Linear Regression model
    model = LinearRegression()

    # Train model
    model.fit(
        X_train,
        y_train
    )

    # Predict BMI
    y_pred = model.predict(
        X_test
    )

    # Calculate R2 score
    r2 = r2_score(
        y_test,
        y_pred
    )

    print("\n" + "-" * 60)
    print(dataset_name)
    print("-" * 60)

    print("Training Samples:", len(X_train))
    print("Testing Samples:", len(X_test))

    print("\nRegression Coefficients:")

    for feature, coefficient in zip(
        features,
        model.coef_
    ):
        print(
            feature,
            ":",
            round(coefficient, 4)
        )

    print(
        "Intercept:",
        round(model.intercept_, 4)
    )

    print(
        "\nR2 Score:",
        round(r2, 4)
    )

    return r2


# ------------------------------------------------------------
# 5. Perform Multiple Regression
# ------------------------------------------------------------

uci_r2 = multiple_regression_analysis(
    uci_diabetes,
    "UCI Diabetes Dataset"
)

pima_r2 = multiple_regression_analysis(
    pima_diabetes,
    "Pima Indians Diabetes Dataset"
)

# ------------------------------------------------------------
# 6. Compare Results
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("MODEL COMPARISON")
print("=" * 70)

comparison = pd.DataFrame({
    "Dataset": [
        "UCI Diabetes Dataset",
        "Pima Indians Diabetes Dataset"
    ],
    "R2 Score": [
        uci_r2,
        pima_r2
    ]
})

print(
    comparison.to_string(
        index=False,
        float_format=lambda x: f"{x:.4f}"
    )
)

# ------------------------------------------------------------
# 7. Interpretation
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("INTERPRETATION")
print("=" * 70)

if uci_r2 > pima_r2:
    print(
        "The UCI working dataset has the higher R2 score."
    )
elif pima_r2 > uci_r2:
    print(
        "The Pima Indians Diabetes Dataset has the higher R2 score."
    )
else:
    print(
        "Both datasets have the same R2 score."
    )

print(
    "\nR2 Score indicates how well Glucose, BloodPressure "
    "and Age explain the variation in BMI."
)

print("\n" + "=" * 70)
print("RESULT")
print("=" * 70)

print(
    "Multiple Regression analysis was performed to predict "
    "BMI using Glucose, BloodPressure and Age."
)

print(
    "The R2 scores were calculated and compared between "
    "the two datasets."
)

print("\n3-C EXPERIMENT COMPLETED SUCCESSFULLY")
print("=" * 70)
```
